## 计算代价
本作业中"代价"这个术语可能有点令人困惑，因为数据本身就是房屋成本。在这里，代价是衡量我们的模型预测房屋目标价格的准确程度。"价格"这个术语用于房屋数据。

单变量代价的方程是：
  $$J(w,b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2 \tag{1}$$ 
 
其中
  $$f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{2}$$
  
- $f_{w,b}(x^{(i)})$ 是我们使用参数 $w,b$ 对第 $i$ 个样本的预测。
- $(f_{w,b}(x^{(i)}) -y^{(i)})^2$ 是目标值与预测值之间的平方差。
- 这些差异在所有 $m$ 个样本上求和，然后除以 `2m` 得到代价 $J(w,b)$。
>注意，在课程中求和范围通常是从1到m，而在代码中是从0到m-1。


下面的代码通过循环每个样本来计算代价。在每次循环中：
- 计算预测值 `f_wb`
- 计算目标值与预测值之间的差并求平方。
- 将其加到总代价中。

```python
def compute_cost(x, y, w, b): 
    """
    计算线性回归的代价函数。
    
    参数：
      x (ndarray (m,)): 数据，m个样本
      y (ndarray (m,)): 目标值
      w,b (标量)      : 模型参数
    
    返回：
        total_cost (float): 使用w,b作为线性回归参数来拟合x和y中数据点的代价
    """
    # 训练样本数量
    m = x.shape[0] 
    
    cost_sum = 0 
    for i in range(m): 
        f_wb = w * x[i] + b   
        cost = (f_wb - y[i]) ** 2  
        cost_sum = cost_sum + cost  
    total_cost = (1 / (2 * m)) * cost_sum  

    return total_cost
```

## 代价函数直觉

<img align="left" src="./images/C1_W1_Lab02_GoalOfRegression.PNG"    style=" width:380px; padding: 10px;  " /> 你的目标是找到一个模型 $f_{w,b}(x) = wx + b$，参数为 $w,b$，能够根据输入 $x$ 准确预测房屋价值。代价是衡量模型在训练数据上准确程度的指标。

上面的代价方程(1)表明，如果可以选择 $w$ 和 $b$ 使得预测值 $f_{w,b}(x)$ 与目标数据 $y$ 匹配，那么 $(f_{w,b}(x^{(i)}) - y^{(i)})^2 $ 项将为零，代价也就最小化了。在这个简单的两点示例中，你可以做到这一点！

在前面的实验中，你确定了 $b=100$ 提供了最优解，所以让我们将 $b$ 设为100，专注于 $w$。

<br/>
下面，使用滑块控件选择使代价最小化的 $w$ 值。图表更新可能需要几秒钟。

```python
plt_intuition(x_train,y_train)
```

该图包含几个值得提及的要点。
- 当 $w = 200$ 时代价最小化，这与前面实验的结果一致
- 因为代价方程中目标值与预测值之间的差被平方了，所以当 $w$ 太大或太小时，代价会急剧增加。
- 使用通过最小化代价选择的 `w` 和 `b` 会得到一条完美拟合数据的直线。
