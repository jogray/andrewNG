```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import SGDRegressor
from sklearn.preprocessing import StandardScaler
from lab_utils_multi import  load_house_data
from lab_utils_common import dlc
np.set_printoptions(precision=2)
plt.style.use('./deeplearning.mplstyle')
```

# 梯度下降
Scikit-learn有一个梯度下降回归模型[sklearn.linear_model.SGDRegressor](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.SGDRegressor.html#examples-using-sklearn-linear-model-sgdregressor)。就像您之前实现的梯度下降一样，这个模型在标准化输入时效果最好。[sklearn.preprocessing.StandardScaler](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.StandardScaler.html#sklearn.preprocessing.StandardScaler)将像之前的实验一样执行z-score标准化。这里它被称为'标准分数'。

### 加载数据集

```python
X_train, y_train = load_house_data()
X_features = ['size(sqft)','bedrooms','floors','age']
```

### 缩放/标准化训练数据

```python
scaler = StandardScaler()
X_norm = scaler.fit_transform(X_train)
print(f"原始X按列的峰峰值范围:{np.ptp(X_train,axis=0)}")   
print(f"标准化后X按列的峰峰值范围:{np.ptp(X_norm,axis=0)}")
```

### 创建并拟合回归模型

```python
sgdr = SGDRegressor(max_iter=1000)
sgdr.fit(X_norm, y_train)
print(sgdr)
print(f"number of iterations completed: {sgdr.n_iter_}, number of weight updates: {sgdr.t_}")
```

### 查看参数
请注意，参数与*标准化*输入数据相关联。拟合参数与之前实验中使用此数据找到的参数非常接近。

```python
b_norm = sgdr.intercept_
w_norm = sgdr.coef_
print(f"model parameters:                   w: {w_norm}, b:{b_norm}")
print( "model parameters from previous lab: w: [110.56 -21.27 -32.71 -37.97], b: 363.16")
```

### 进行预测
预测训练数据的目标值。同时使用`predict`例程和使用$w$和$b$计算。

```python
# 使用sgdr.predict()进行预测
y_pred_sgd = sgdr.predict(X_norm)
# 使用w,b进行预测
y_pred = np.dot(X_norm, w_norm) + b_norm  
print(f"使用np.dot()和sgdr.predict的预测匹配: {(y_pred == y_pred_sgd).all()}")

print(f"训练集上的预测:\n{y_pred[:4]}" )
print(f"目标值 \n{y_train[:4]}")
```

### 绘制结果
让我们绘制预测值与目标值的关系图。

```python
# 绘制预测值和目标值与原始特征的关系图    
fig,ax=plt.subplots(1,4,figsize=(12,3),sharey=True)
for i in range(len(ax)):
    ax[i].scatter(X_train[:,i],y_train, label = 'target')
    ax[i].set_xlabel(X_features[i])
    ax[i].scatter(X_train[:,i],y_pred,color=dlc["dlorange"], label = 'predict')
ax[0].set_ylabel("Price"); ax[0].legend();
fig.suptitle("使用z-score标准化模型的目标值与预测值对比")
plt.show()
```
