# 线性回归，闭式解
Scikit-learn有[线性回归模型](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html#sklearn.linear_model.LinearRegression)，它实现了闭式线性回归。

让我们使用早期实验中的数据 - 一栋1000平方英尺的房屋售价为\$300,000，一栋2000平方英尺的房屋售价为\$500,000。

| 面积（1000平方英尺）     | 价格（千美元） |
| ----------------| ------------------------ |
| 1               | 300                      |
| 2               | 500                      |


### 加载数据集

```python
X_train = np.array([1.0, 2.0])   #特征
y_train = np.array([300, 500])   #目标值
```

### 创建并拟合模型
下面的代码使用scikit-learn执行回归。
第一步创建一个回归对象。
第二步使用与对象关联的方法之一`fit`。这执行回归，将参数拟合到输入数据。工具包期望二维X矩阵。

```python
linear_model = LinearRegression()
# X必须是2-D矩阵
linear_model.fit(X_train.reshape(-1, 1), y_train) 
```

### 查看参数 
在scikit-learn中，$\mathbf{w}$和$\mathbf{b}$参数被称为'系数'和'截距'。

```python
b = linear_model.intercept_
w = linear_model.coef_
print(f"w = {w:}, b = {b:0.2f}")
print(f"'manual' prediction: f_wb = wx+b : {1200*w + b}")
```

### 进行预测

调用`predict`函数生成预测。

```python
y_pred = linear_model.predict(X_train.reshape(-1, 1))

print("训练集上的预测:", y_pred)

X_test = np.array([[1200]])
print(f"1200平方英尺房屋的预测: ${linear_model.predict(X_test)[0]:0.2f}")
```

## 第二个示例
第二个示例来自之前具有多个特征的实验。最终参数值和预测与该实验中未标准化的'长期运行'结果非常接近。那个未标准化的运行花了几个小时才产生结果，而这几乎是瞬时的。闭式解在较小的数据集上效果很好，但在较大的数据集上可能计算量很大。
>闭式解不需要标准化。

```python
# 加载数据集
X_train, y_train = load_house_data()
X_features = ['size(sqft)','bedrooms','floors','age']
```

```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train) 
```

```python
b = linear_model.intercept_
w = linear_model.coef_
print(f"w = {w:}, b = {b:0.2f}")
```

```python
print(f"训练集上的预测:\n {linear_model.predict(X_train)[:4]}" )
print(f"使用w,b的预测:\n {(X_train @ w + b)[:4]}")
print(f"目标值 \n {y_train[:4]}")

x_house = np.array([1200, 3,1, 40]).reshape(-1,4)
x_house_predict = linear_model.predict(x_house)[0]
print(f" 1200平方英尺、3卧室、1楼层、40年房龄的房屋预测价格 = ${x_house_predict*1000:0.2f}")
```
