## 数据集

假设你有以下训练数据集

```python
X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1]).reshape(-1,1) 
```

让我们使用辅助函数来绘制这些数据。标签为 $y=1$ 的数据点显示为红色十字，而标签为 $y=0$ 的数据点显示为蓝色圆圈。

```python
fig,ax = plt.subplots(1,1,figsize=(4,4))
plot_data(X, y, ax)

ax.axis([0, 4, 0, 3.5])
ax.set_ylabel('$x_1$')
ax.set_xlabel('$x_0$')
plt.show()
```

## 逻辑回归模型


* 假设你想在这份数据上训练一个逻辑回归模型，其形式为

  $f(x) = g(w_0x_0+w_1x_1 + b)$
  
  其中 $g(z) = \frac{1}{1+e^{-z}}$，即sigmoid函数


* 假设你训练了模型并得到参数 $b = -3, w_0 = 1, w_1 = 1$。也就是说，

  $f(x) = g(x_0+x_1-3)$

  （你将在课程的后续部分学习如何将这些参数拟合到数据）
  
  
让我们尝试通过绘制决策边界来理解这个训练好的模型在预测什么

### 逻辑回归和决策边界复习

* 回忆一下，对于逻辑回归，模型表示为

  $$f_{\mathbf{w},b}(\mathbf{x}^{(i)}) = g(\mathbf{w} \cdot \mathbf{x}^{(i)} + b) \tag{1}$$

  其中 $g(z)$ 被称为sigmoid函数，它将所有输入值映射到0和1之间的值：

  $g(z) = \frac{1}{1+e^{-z}}\tag{2}$
  且 $\mathbf{w} \cdot \mathbf{x}$ 是向量点积：
  
  $$\mathbf{w} \cdot \mathbf{x} = w_0 x_0 + w_1 x_1$$
  
  
 * 我们将模型的输出（$f_{\mathbf{w},b}(x)$）解释为给定 $\mathbf{x}$ 并以 $\mathbf{w}$ 和 $b$ 为参数的 $y=1$ 的概率。
* 因此，要从逻辑回归模型获得最终预测（$y=0$ 或 $y=1$），我们可以使用以下启发式方法 -

  如果 $f_{\mathbf{w},b}(x) >= 0.5$，预测 $y=1$
  
  如果 $f_{\mathbf{w},b}(x) < 0.5$，预测 $y=0$
  
  
* 让我们绘制sigmoid函数，看看 $g(z) >= 0.5$ 在哪里

```python
# 在-10到10的值范围内绘制sigmoid(z)
z = np.arange(-10,11)

fig,ax = plt.subplots(1,1,figsize=(5,3))
# 绘制z与sigmoid(z)的关系图
ax.plot(z, sigmoid(z), c="b")

ax.set_title("Sigmoid function")
ax.set_ylabel('sigmoid(z)')
ax.set_xlabel('z')
draw_vthresh(ax,0)
```

* 如你所见，当 $z >=0$ 时，$g(z) >= 0.5$

* 对于逻辑回归模型，$z = \mathbf{w} \cdot \mathbf{x} + b$。因此，

  如果 $\mathbf{w} \cdot \mathbf{x} + b >= 0$，模型预测 $y=1$
  
  如果 $\mathbf{w} \cdot \mathbf{x} + b < 0$，模型预测 $y=0$
  
  
  
### 绘制决策边界

现在，让我们回到我们的例子，理解逻辑回归模型是如何进行预测的。

* 我们的逻辑回归模型的形式为

  $f(\mathbf{x}) = g(-3 + x_0+x_1)$


* 从你上面学到的内容，你可以看到这个模型在 $-3 + x_0+x_1 >= 0$ 时预测 $y=1$

让我们看看这在图形上是什么样的。我们将从绘制 $-3 + x_0+x_1 = 0$ 开始，这等价于 $x_1 = 3 - x_0$。


```python
# 选择0到6之间的值
x0 = np.arange(0,6)

x1 = 3 - x0
fig,ax = plt.subplots(1,1,figsize=(5,4))
# 绘制决策边界
ax.plot(x0,x1, c="b")
ax.axis([0, 4, 0, 3.5])

# 填充线下方的区域
ax.fill_between(x0,x1, alpha=0.2)

# 绘制原始数据
plot_data(X,y,ax)
ax.set_ylabel(r'$x_1$')
ax.set_xlabel(r'$x_0$')
plt.show()
```

* 在上图中，蓝线代表直线 $x_0 + x_1 - 3 = 0$，它应该与x1轴相交于3（如果我们设置 $x_1$ = 3, $x_0$ = 0），与x0轴相交于3（如果我们设置 $x_1$ = 0, $x_0$ = 3）。


* 阴影区域代表 $-3 + x_0+x_1 < 0$。线上方的区域是 $-3 + x_0+x_1 > 0$。


* 阴影区域（线下方）中的任何点都被分类为 $y=0$。线上或线上方的任何点都被分类为 $y=1$。这条线被称为"决策边界"。

正如我们在讲座中看到的，通过使用更高阶的多项式项（例如：$f(x) = g( x_0^2 + x_1 -1)$），我们可以得到更复杂的非线性边界。
