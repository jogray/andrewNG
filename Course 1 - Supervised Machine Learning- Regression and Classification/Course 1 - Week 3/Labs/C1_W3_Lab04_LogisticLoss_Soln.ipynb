## 逻辑回归的平方误差？
<img align="left" src="./images/C1_W3_SqErrorVsLogistic.png"     style=" width:400px; padding: 10px; " > 回忆一下，对于**线性**回归，我们使用了**平方误差成本函数**：
单变量平方误差成本的方程是：
$$J(w,b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2 \tag{1}
$$ 
 
其中
$$
f_{w,b}(x^{(i)}) = wx^{(i)} + b \tag{2}
$$


回忆一下，平方误差成本有一个很好的特性，即沿着成本的导数可以找到最小值。

```python
soup_bowl()
```

这个成本函数对线性回归很有效，很自然地也会考虑将其用于逻辑回归。然而，正如上面的幻灯片指出的，$f_{wb}(x)$ 现在有一个非线性组件，即sigmoid函数：$f_{w,b}(x^{(i)}) = sigmoid(wx^{(i)} + b )$。让我们在之前的实验室示例上尝试平方误差成本，现在包括sigmoid。

这是我们的训练数据：

```python
x_train = np.array([0., 1, 2, 3, 4, 5],dtype=np.longdouble)
y_train = np.array([0,  0, 0, 1, 1, 1],dtype=np.longdouble)
plt_simple_example(x_train, y_train)
```

现在，让我们使用*平方误差成本*来获取成本的表面图：
  $$J(w,b) = \frac{1}{2m} \sum\limits_{i = 0}^{m-1} (f_{w,b}(x^{(i)}) - y^{(i)})^2 
  $$ 
 
其中
  $$f_{w,b}(x^{(i)}) = sigmoid(wx^{(i)} + b )$$


```python
plt.close('all')
plt_logistic_squared_error(x_train,y_train)
plt.show()
```

虽然这产生了一个非常有趣的图，但上面的表面远不如线性回归中的'汤碗'平滑！

逻辑回归需要一个更适合其非线性性质的成本函数。这从一个损失函数开始。下面将描述这一点。

## 逻辑损失函数

逻辑回归使用一个更适合分类任务的损失函数，其中目标是0或1而不是任何数字。

>**定义说明：** 在本课程中，使用以下定义：
**损失**是单个示例与其目标值之间差异的度量，而
**成本**是训练集上损失的度量


定义如下：
* $loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)})$ 是单个数据点的成本，即：

\begin{equation}
  loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)}) = \begin{cases}
    - \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) & \text{if $y^{(i)}=1$}\\
    - \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) & \text{if $y^{(i)}=0$}
  \end{cases}
\end{equation}


*  $f_{\mathbf{w},b}(\mathbf{x}^{(i)})$ 是模型的预测，而 $y^{(i)}$ 是目标值。

*  $f_{\mathbf{w},b}(\mathbf{x}^{(i)}) = g(\mathbf{w} \cdot\mathbf{x}^{(i)}+b)$，其中函数 $g$ 是sigmoid函数。

这个损失函数的定义特征是它使用两条独立的曲线。一条用于目标为零（$y=0$）的情况，另一条用于目标为一（$y=1$）的情况。结合起来，这些曲线提供了对损失函数有用的行为，即当预测与目标匹配时为零，当预测与目标不同时值迅速增加。考虑下面的曲线：

```python
plt_two_logistic_loss_curves()
```

结合起来，这些曲线类似于平方误差损失的二次曲线。注意，x轴是 $f_{\mathbf{w},b}$，它是sigmoid的输出。sigmoid输出严格在0和1之间。

上面的损失函数可以重写以便更容易实现。
    $$loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)}) = (-y^{(i)} \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - y^{(i)}\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right)$$
  
这是一个看起来相当复杂的等式。但当你考虑到 $y^{(i)}$ 只能取两个值（0 和 1）时，它就没那么令人望而生畏了。我们可以分两种情况来考虑这个等式：  
当 $ y^{(i)} = 0$ 时，左边的项被消除：
$$
\begin{align}
loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), 0) &= (-(0) \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - 0\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) \\
&= -\log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right)
\end{align}
$$
当 $ y^{(i)} = 1$ 时，右边的项被消除：
$$
\begin{align}
  loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), 1) &=  (-(1) \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - 1\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right)\\
  &=  -\log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right)
\end{align}
$$

好的，有了这个新的逻辑损失函数，就可以产生一个包含所有样本损失的代价函数。这将是下一个实验的主题。现在，让我们来看看上面所考虑的简单示例的代价与参数曲线：

```python
plt.close('all')
cst = plt_logistic_cost(x_train,y_train)
```

这条曲线非常适合梯度下降！它没有平台期、局部最小值或不连续点。请注意，它不像平方误差那样是一个碗形。图中同时绘制了代价和代价的对数，以说明当代价较小时，曲线仍然具有斜率并继续下降。提示：你可以使用鼠标旋转上面的图表。
