## 代价函数

在之前的实验中，你开发了*逻辑损失*函数。回顾一下，损失是针对单个样本定义的。在这里，你将各个损失组合成**代价**，它包含所有样本。


回顾一下，逻辑回归的代价函数形式为

$$ J(\mathbf{w},b) = \frac{1}{m} \sum_{i=0}^{m-1} \left[ loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)}) \right] \tag{1}$$

其中
* $loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)})$ 是单个数据点的代价，其表达式为：

    $$loss(f_{\mathbf{w},b}(\mathbf{x}^{(i)}), y^{(i)}) = -y^{(i)} \log\left(f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) - \left( 1 - y^{(i)}\right) \log \left( 1 - f_{\mathbf{w},b}\left( \mathbf{x}^{(i)} \right) \right) \tag{2}$$
    
*  其中 m 是数据集中训练样本的数量，并且：
$$
\begin{align}
  f_{\mathbf{w},b}(\mathbf{x^{(i)}}) &= g(z^{(i)})\tag{3} \\
  z^{(i)} &= \mathbf{w} \cdot \mathbf{x}^{(i)}+ b\tag{4} \\
  g(z^{(i)}) &= \frac{1}{1+e^{-z^{(i)}}}\tag{5} 
\end{align}
$$

#### 代码说明

`compute_cost_logistic` 的算法遍历所有样本，计算每个样本的损失并累加。

注意，变量 X 和 y 不是标量值，而是形状分别为 ($m, n$) 和 ($𝑚$,) 的矩阵，其中 $𝑛$ 是特征数量，$𝑚$ 是训练样本数量。


```python
def compute_cost_logistic(X, y, w, b):
    """
    计算代价

    参数:
      X (ndarray (m,n)): 数据，m个样本，每个样本有n个特征
      y (ndarray (m,)) : 目标值
      w (ndarray (n,)) : 模型参数
      b (标量)       : 模型参数
      
    返回:
      cost (标量): 代价
    """

    m = X.shape[0]
    cost = 0.0
    for i in range(m):
        z_i = np.dot(X[i],w) + b
        f_wb_i = sigmoid(z_i)
        cost +=  -y[i]*np.log(f_wb_i) - (1-y[i])*np.log(1-f_wb_i)
             
    cost = cost / m
    return cost
```

使用下面的单元格检查代价函数的实现。

```python
w_tmp = np.array([1,1])
b_tmp = -3
print(compute_cost_logistic(X_train, y_train, w_tmp, b_tmp))
```

**期望输出**: 0.3668667864055175

## 示例
现在，让我们看看对于不同的 $w$ 值，代价函数的输出是什么。

* 在之前的实验中，你绘制了 $b = -3, w_0 = 1, w_1 = 1$ 的决策边界。也就是说，你有 `w = np.array([-3,1,1])`。

* 假设你想看看 $b = -4, w_0 = 1, w_1 = 1$，即 `w = np.array([-4,1,1])` 是否能提供更好的模型。

让我们先绘制这两个不同 $b$ 值的决策边界，看看哪个更拟合数据。

* 对于 $b = -3, w_0 = 1, w_1 = 1$，我们将绘制 $-3 + x_0+x_1 = 0$（蓝色显示）
* 对于 $b = -4, w_0 = 1, w_1 = 1$，我们将绘制 $-4 + x_0+x_1 = 0$（品红色显示）

```python
import matplotlib.pyplot as plt

# 选择 0 到 6 之间的值
x0 = np.arange(0,6)

# 绘制两个决策边界
x1 = 3 - x0
x1_other = 4 - x0

fig,ax = plt.subplots(1, 1, figsize=(4,4))
# 绘制决策边界
ax.plot(x0,x1, c=dlc["dlblue"], label="$b$=-3")
ax.plot(x0,x1_other, c=dlc["dlmagenta"], label="$b$=-4")
ax.axis([0, 4, 0, 4])

# 绘制原始数据
plot_data(X_train,y_train,ax)
ax.axis([0, 4, 0, 4])
ax.set_ylabel('$x_1$', fontsize=12)
ax.set_xlabel('$x_0$', fontsize=12)
plt.legend(loc="upper right")
plt.title("决策边界")
plt.show()
```

从这个图中你可以看到 `w = np.array([-4,1,1])` 对训练数据来说是一个更差的模型。让我们看看代价函数的实现是否反映了这一点。

```python
w_array1 = np.array([1,1])
b_1 = -3
w_array2 = np.array([1,1])
b_2 = -4

print("Cost for b = -3 : ", compute_cost_logistic(X_train, y_train, w_array1, b_1))
print("Cost for b = -4 : ", compute_cost_logistic(X_train, y_train, w_array2, b_2))
```

**期望输出**

Cost for b = -3 :  0.3668667864055175

Cost for b = -4 :  0.5036808636748461


你可以看到代价函数的行为符合预期，`w = np.array([-4,1,1])` 的代价确实高于 `w = np.array([-3,1,1])` 的代价。
